```python

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.linear_model import Ridge
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

from xgboost import XGBRegressor

# --------------------------------------------------
# 1. DEFINE TARGET AND PREDICTORS
# --------------------------------------------------

TARGET = "Yield (Kg/ha)"

# Climate + district + year.
# Area and Production are deliberately excluded.
# Irrigation and PMFBY are reserved for later experiments.

numeric_features = [
    "year",
    "rainfall_annual_mm",
    "rainfall_may_mm",
    "rainfall_june_mm",
    "rainfall_may_june_mm",
    "temperature_annual_c",
    "temperature_may_c",
    "temperature_june_c",
    "temperature_may_june_c"
]

categorical_features = ["district"]
features = numeric_features + categorical_features

# Keep rows with a target and required predictor values.
data = df.dropna(subset=[TARGET] + features).copy()
data = data.sort_values(["year", "district"]).reset_index(drop=True)

X = data[features]
y = data[TARGET]

print("Rows used:", len(data))
print("Districts:", data["district"].nunique())
print("Years:", data["year"].min(), "-", data["year"].max())

# --------------------------------------------------
# 2. PREPROCESSING
# --------------------------------------------------

# Missing numeric predictors, if any, are imputed using
# training data only. District names are one-hot encoded.
preprocessor = ColumnTransformer([
    ("numeric", Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]), numeric_features),

    ("district", Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(handle_unknown="ignore"))
    ]), categorical_features)
])

# --------------------------------------------------
# 3. DEFINE THE THREE MODELS
# --------------------------------------------------

def make_models():
    return {
        "Ridge": Ridge(alpha=1.0),

        "Random Forest": RandomForestRegressor(
            n_estimators=300,
            random_state=42,
            n_jobs=-1
        ),

        "XGBoost": XGBRegressor(
            n_estimators=300,
            max_depth=3,
            learning_rate=0.05,
            objective="reg:squarederror",
            random_state=42,
            n_jobs=-1
        )
    }

def make_pipeline(model):
    return Pipeline([
        ("preprocessing", preprocessor),
        ("model", model)
    ])

def calculate_metrics(actual, predicted):
    return {
        "R2": r2_score(actual, predicted),
        "MAE": mean_absolute_error(actual, predicted),
        "RMSE": np.sqrt(mean_squared_error(actual, predicted))
    }

# --------------------------------------------------
# 4. EXPERIMENT 1: RANDOM 70/30 SPLIT
# --------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.30,
    random_state=42,
    shuffle=True
)

random_results = []
random_predictions = []

for name, model in make_models().items():
    pipe = make_pipeline(model)
    pipe.fit(X_train, y_train)
    predicted = pipe.predict(X_test)

    random_predictions.append(pd.DataFrame({
        "district": data.loc[X_test.index, "district"].values,
        "year": data.loc[X_test.index, "year"].values,
        "actual_yield": y_test.values,
        "predicted_yield": predicted,
        "model": name
    }))

    # Full random test-set metrics
    scores = calculate_metrics(y_test, predicted)
    random_results.append({
        "experiment": "Random 70/30 - all test years",
        "model": name,
        **scores
    })

    # Comparable-period metrics: years 2005 onward
    test_years = data.loc[X_test.index, "year"].values
    mask = test_years >= 2005

    if mask.sum() > 1:
        scores_period = calculate_metrics(
            y_test.to_numpy()[mask],
            predicted[mask]
        )
        random_results.append({
            "experiment": "Random 70/30 - test years 2005 onward",
            "model": name,
            **scores_period
        })

# --------------------------------------------------
# 5. EXPERIMENT 2: WALK-FORWARD VALIDATION
# --------------------------------------------------

# Start testing in 2005, after five calendar years
# (2000-2004) are available for initial training.
walk_predictions = []
walk_results = []

for test_year in sorted(data["year"].unique()):
    if test_year < 2005:
        continue

    train_idx = data.index[data["year"] < test_year]
    test_idx = data.index[data["year"] == test_year]

    if len(train_idx) == 0 or len(test_idx) == 0:
        continue

    X_train_year = X.loc[train_idx]
    y_train_year = y.loc[train_idx]
    X_test_year = X.loc[test_idx]
    y_test_year = y.loc[test_idx]

    for name, model in make_models().items():
        pipe = make_pipeline(model)
        pipe.fit(X_train_year, y_train_year)
        predicted = pipe.predict(X_test_year)

        walk_predictions.append(pd.DataFrame({
            "district": data.loc[test_idx, "district"].values,
            "year": data.loc[test_idx, "year"].values,
            "actual_yield": y_test_year.values,
            "predicted_yield": predicted,
            "model": name
        }))

# Combine yearly predictions and calculate pooled metrics
walk_predictions = pd.concat(walk_predictions, ignore_index=True)

for name in make_models().keys():
    part = walk_predictions[walk_predictions["model"] == name]
    scores = calculate_metrics(
        part["actual_yield"],
        part["predicted_yield"]
    )

    walk_results.append({
        "experiment": "Walk-forward - pooled test years",
        "model": name,
        **scores
    })

# --------------------------------------------------
# 6. SAVE RESULTS
# --------------------------------------------------

random_predictions = pd.concat(random_predictions, ignore_index=True)

results = pd.DataFrame(random_results + walk_results)

results.to_csv("model_metrics.csv", index=False)
random_predictions.to_csv("random_split_predictions.csv", index=False)
walk_predictions.to_csv("walk_forward_predictions.csv", index=False)

print("\nMODEL METRICS")
print(results.round(3).to_string(index=False))

print("\nSaved files:")
print("model_metrics.csv")
print("random_split_predictions.csv")
print("walk_forward_predictions.csv")
```

```python

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.linear_model import Ridge
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
from xgboost import XGBRegressor

# =========================================================
# 1. SETTINGS
# =========================================================

TARGET = "Yield (Kg/ha)"
RANDOM_STATE = 42
FIRST_TEST_YEAR = 2005

climate_features = [
    "rainfall_annual_mm",
    "rainfall_may_mm",
    "rainfall_june_mm",
    "rainfall_may_june_mm",
    "temperature_annual_c",
    "temperature_may_c",
    "temperature_june_c",
    "temperature_may_june_c"
]

# Experiment A: include year
features_with_year = ["year"] + climate_features

# Experiment B: exclude year
features_without_year = climate_features.copy()

categorical_features = ["district"]


# =========================================================
# 2. PREPARE DATA
# =========================================================

required_columns = (
    [TARGET, "district", "year"]
    + climate_features
)

data = df.dropna(subset=required_columns).copy()
data["year"] = data["year"].astype(int)
data = data.sort_values(["year", "district"]).reset_index(drop=True)

print("Rows used:", len(data))
print("Districts:", data["district"].nunique())
print("Years:", data["year"].min(), "-", data["year"].max())


# =========================================================
# 3. MODELS AND PREPROCESSING
# =========================================================

def make_models():
    return {
        "Ridge": Ridge(alpha=1.0),
        "Random Forest": RandomForestRegressor(
            n_estimators=300,
            random_state=RANDOM_STATE,
            n_jobs=-1
        ),
        "XGBoost": XGBRegressor(
            n_estimators=300,
            max_depth=3,
            learning_rate=0.05,
            objective="reg:squarederror",
            random_state=RANDOM_STATE,
            n_jobs=-1
        )
    }

def make_pipeline(model, numeric_features):
    preprocessor = ColumnTransformer([
        ("numeric", Pipeline([
            ("imputer", SimpleImputer(strategy="median")),
            ("scaler", StandardScaler())
        ]), numeric_features),

        ("district", Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("encoder", OneHotEncoder(handle_unknown="ignore"))
        ]), categorical_features)
    ])

    return Pipeline([
        ("preprocessing", preprocessor),
        ("model", model)
    ])

def get_metrics(actual, predicted):
    return {
        "R2": r2_score(actual, predicted),
        "MAE": mean_absolute_error(actual, predicted),
        "RMSE": np.sqrt(mean_squared_error(actual, predicted))
    }


# =========================================================
# 4. RUN BOTH EXPERIMENTS WITH AND WITHOUT YEAR
# =========================================================

all_predictions = []
all_metrics = []

feature_versions = {
    "With year": features_with_year,
    "Without year": features_without_year
}

for feature_version, numeric_features in feature_versions.items():

    features = numeric_features + categorical_features
    X = data[features]
    y = data[TARGET]

    # Same random split seed for both feature versions
    X_train, X_test, y_train, y_test = train_test_split(
        X, y,
        test_size=0.30,
        random_state=RANDOM_STATE,
        shuffle=True
    )

    # Keep metadata aligned with test rows
    random_test_meta = data.loc[X_test.index, ["district", "year"]]

    for model_name, model in make_models().items():

        # ---- Random 70/30 split ----
        pipe = make_pipeline(model, numeric_features)
        pipe.fit(X_train, y_train)
        pred = pipe.predict(X_test)

        random_df = pd.DataFrame({
            "district": random_test_meta["district"].to_numpy(),
            "year": random_test_meta["year"].to_numpy(),
            "actual_yield": y_test.to_numpy(),
            "predicted_yield": pred,
            "model": model_name,
            "feature_version": feature_version,
            "validation": "Random split"
        })

        all_predictions.append(random_df)

        # Overall random test performance
        all_metrics.append({
            "feature_version": feature_version,
            "model": model_name,
            "validation": "Random split - all test years",
            **get_metrics(y_test, pred)
        })

        # Random test rows restricted to 2005 onward
        recent_mask = random_test_meta["year"].to_numpy() >= FIRST_TEST_YEAR

        all_metrics.append({
            "feature_version": feature_version,
            "model": model_name,
            "validation": "Random split - 2005 onward",
            **get_metrics(
                y_test.to_numpy()[recent_mask],
                pred[recent_mask]
            )
        })

        # ---- Walk-forward validation ----
        walk_parts = []

        for test_year in sorted(data["year"].unique()):
            if test_year < FIRST_TEST_YEAR:
                continue

            train_idx = data.index[data["year"] < test_year]
            test_idx = data.index[data["year"] == test_year]

            if len(train_idx) == 0 or len(test_idx) == 0:
                continue

            walk_pipe = make_pipeline(model, numeric_features)
            walk_pipe.fit(
                data.loc[train_idx, features],
                data.loc[train_idx, TARGET]
            )

            year_pred = walk_pipe.predict(data.loc[test_idx, features])

            walk_parts.append(pd.DataFrame({
                "district": data.loc[test_idx, "district"].to_numpy(),
                "year": data.loc[test_idx, "year"].to_numpy(),
                "actual_yield": data.loc[test_idx, TARGET].to_numpy(),
                "predicted_yield": year_pred,
                "model": model_name,
                "feature_version": feature_version,
                "validation": "Walk-forward"
            }))

        walk_df = pd.concat(walk_parts, ignore_index=True)
        all_predictions.append(walk_df)

        all_metrics.append({
            "feature_version": feature_version,
            "model": model_name,
            "validation": "Walk-forward - all test years",
            **get_metrics(
                walk_df["actual_yield"],
                walk_df["predicted_yield"]
            )
        })


# =========================================================
# 5. MATCH THE EXACT SAME DISTRICT-YEAR TEST OBSERVATIONS
# =========================================================

predictions = pd.concat(all_predictions, ignore_index=True)
metrics = pd.DataFrame(all_metrics)

matched_results = []

for feature_version in feature_versions:
    for model_name in make_models().keys():

        random_part = predictions[
            (predictions["feature_version"] == feature_version)
            & (predictions["model"] == model_name)
            & (predictions["validation"] == "Random split")
            & (predictions["year"] >= FIRST_TEST_YEAR)
        ].copy()

        walk_part = predictions[
            (predictions["feature_version"] == feature_version)
            & (predictions["model"] == model_name)
            & (predictions["validation"] == "Walk-forward")
        ].copy()

        # Match using district and year so both methods are
        # scored on the same actual yield observations.
        matched = random_part.merge(
            walk_part,
            on=["district", "year"],
            how="inner",
            suffixes=("_random", "_walk")
        )

        if len(matched) == 0:
            print("No matched observations:", feature_version, model_name)
            continue

        actual = matched["actual_yield_random"]

        matched_results.append({
            "feature_version": feature_version,
            "model": model_name,
            "validation": "Matched random split",
            "matched_rows": len(matched),
            **get_metrics(actual, matched["predicted_yield_random"])
        })

        matched_results.append({
            "feature_version": feature_version,
            "model": model_name,
            "validation": "Matched walk-forward",
            "matched_rows": len(matched),
            **get_metrics(actual, matched["predicted_yield_walk"])
        })

matched_metrics = pd.DataFrame(matched_results)


# =========================================================
# 6. DISPLAY AND SAVE RESULTS
# =========================================================

print("\nALL METRICS")
print(metrics.round(3).to_string(index=False))

print("\nMATCHED OBSERVATION COMPARISON")
print(matched_metrics.round(3).to_string(index=False))

metrics.to_csv("sensitivity_all_metrics.csv", index=False)
matched_metrics.to_csv("sensitivity_matched_metrics.csv", index=False)
predictions.to_csv("sensitivity_predictions.csv", index=False)

print("\nSaved:")
print("sensitivity_all_metrics.csv")
print("sensitivity_matched_metrics.csv")
print("sensitivity_predictions.csv")
```

```python

import pandas as pd
import numpy as np
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

# =========================================================
# 1. CHECK THAT THE REQUIRED DATA IS AVAILABLE
# =========================================================

required = [
    "district", "year", "irrigation_group",
    "irrigation_source_year", "Yield (Kg/ha)"
]

missing = [c for c in required if c not in df.columns]
if missing:
    raise ValueError(f"Missing columns in df: {missing}")

required_predictions = [
    "district", "year", "actual_yield", "predicted_yield",
    "model", "feature_version", "validation"
]

missing = [c for c in required_predictions if c not in predictions.columns]
if missing:
    raise ValueError(f"Missing columns in predictions: {missing}")


# =========================================================
# 2. KEEP ONLY GENUINELY OBSERVED IRRIGATION CATEGORIES
# =========================================================

irrigation = df[
    ["district", "year", "irrigation_group",
     "irrigation_source_year"]
].copy()

irrigation["year"] = pd.to_numeric(irrigation["year"], errors="coerce")
irrigation["irrigation_source_year"] = pd.to_numeric(
    irrigation["irrigation_source_year"], errors="coerce"
)

# A row is included only when its irrigation category is present
# and its source reporting year matches the panel year.
valid_groups = ["Irrigated", "Mixed", "Rain-fed"]

irrigation["irrigation_group"] = (
    irrigation["irrigation_group"].astype("string").str.strip()
)

observed_irrigation = irrigation[
    irrigation["irrigation_group"].isin(valid_groups)
    & irrigation["irrigation_source_year"].notna()
    & (irrigation["irrigation_source_year"] == irrigation["year"])
].copy()

# Ensure there is at most one irrigation category per district-year.
if observed_irrigation.duplicated(["district", "year"]).any():
    raise ValueError(
        "Duplicate observed irrigation district-year rows found. "
        "Resolve these before analysis."
    )

print("Observed irrigation district-year records:",
      len(observed_irrigation))
print("\nObserved categories:")
print(observed_irrigation["irrigation_group"].value_counts())


# =========================================================
# 3. JOIN OBSERVED IRRIGATION TO MODEL PREDICTIONS
# =========================================================

pred = predictions.copy()
pred["year"] = pd.to_numeric(pred["year"], errors="coerce")

irrigation_predictions = pred.merge(
    observed_irrigation[["district", "year", "irrigation_group"]],
    on=["district", "year"],
    how="inner",
    validate="many_to_one"
)

if irrigation_predictions.empty:
    raise ValueError(
        "No predictions matched observed irrigation records. "
        "Check district names and year formats."
    )

# Error = predicted yield minus actual yield.
# Positive means overprediction; negative means underprediction.
irrigation_predictions["prediction_error"] = (
    irrigation_predictions["predicted_yield"]
    - irrigation_predictions["actual_yield"]
)

irrigation_predictions["absolute_error"] = (
    irrigation_predictions["prediction_error"].abs()
)


# =========================================================
# 4. METRICS BY IRRIGATION GROUP
# =========================================================

def subgroup_metrics(group):
    actual = group["actual_yield"].to_numpy()
    predicted = group["predicted_yield"].to_numpy()

    return pd.Series({
        "n_observations": len(group),
        "n_districts": group["district"].nunique(),
        "R2": r2_score(actual, predicted) if len(group) >= 2 else np.nan,
        "MAE_kg_ha": mean_absolute_error(actual, predicted),
        "RMSE_kg_ha": np.sqrt(mean_squared_error(actual, predicted)),
        "Mean_error_kg_ha": group["prediction_error"].mean()
    })

grouped = (
    irrigation_predictions
    .groupby(
        ["feature_version", "model", "validation", "irrigation_group"],
        observed=True
    )
    .apply(subgroup_metrics, include_groups=False)
    .reset_index()
)

print("\nMETRICS BY IRRIGATION GROUP")
print(grouped.round(3).to_string(index=False))


# =========================================================
# 5. FAIRER COMPARISON: SAME DISTRICT-YEARS
# =========================================================

matched_results = []

for feature_version in pred["feature_version"].dropna().unique():
    for model_name in pred["model"].dropna().unique():

        random_part = pred[
            (pred["feature_version"] == feature_version)
            & (pred["model"] == model_name)
            & (pred["validation"] == "Random split")
            & (pred["year"] >= 2005)
        ].copy()

        walk_part = pred[
            (pred["feature_version"] == feature_version)
            & (pred["model"] == model_name)
            & (pred["validation"] == "Walk-forward")
        ].copy()

        matched = random_part.merge(
            walk_part,
            on=["district", "year"],
            how="inner",
            suffixes=("_random", "_walk")
        )

        # Attach only observed irrigation categories.
        matched = matched.merge(
            observed_irrigation[["district", "year", "irrigation_group"]],
            on=["district", "year"],
            how="inner",
            validate="many_to_one"
        )

        for group_name, part in matched.groupby("irrigation_group"):
            for method, actual_col, pred_col in [
                ("Random split", "actual_yield_random",
                 "predicted_yield_random"),
                ("Walk-forward", "actual_yield_walk",
                 "predicted_yield_walk")
            ]:
                actual = part[actual_col].to_numpy()
                predicted = part[pred_col].to_numpy()

                matched_results.append({
                    "feature_version": feature_version,
                    "model": model_name,
                    "validation": method,
                    "irrigation_group": group_name,
                    "matched_observations": len(part),
                    "n_districts": part["district"].nunique(),
                    "R2": (
                        r2_score(actual, predicted)
                        if len(part) >= 2 else np.nan
                    ),
                    "MAE_kg_ha": mean_absolute_error(actual, predicted),
                    "RMSE_kg_ha": np.sqrt(
                        mean_squared_error(actual, predicted)
                    )
                })

matched_group_metrics = pd.DataFrame(matched_results)

print("\nMATCHED RANDOM VS WALK-FORWARD BY IRRIGATION GROUP")
if matched_group_metrics.empty:
    print("No matched observations were available for this comparison.")
else:
    print(matched_group_metrics.round(3).to_string(index=False))


# =========================================================
# 6. SAVE RESULTS
# =========================================================

irrigation_predictions.to_csv(
    "irrigation_subgroup_predictions.csv", index=False
)

grouped.to_csv(
    "irrigation_subgroup_metrics.csv", index=False
)

matched_group_metrics.to_csv(
    "irrigation_matched_comparison.csv", index=False
)

print("\nSaved files:")
print("irrigation_subgroup_predictions.csv")
print("irrigation_subgroup_metrics.csv")
print("irrigation_matched_comparison.csv")
```

```python

# ============================================================
# RECREATE BASELINE MODEL PREDICTIONS
# ============================================================

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.linear_model import Ridge
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

# XGBoost
try:
    from xgboost import XGBRegressor
except ImportError:
    !pip -q install xgboost
    from xgboost import XGBRegressor

# ------------------------------------------------------------
# 1. SET UP DATA
# ------------------------------------------------------------

target = "Yield (Kg/ha)"

numeric_features = [
    "year",
    "rainfall_annual_mm",
    "rainfall_may_mm",
    "rainfall_june_mm",
    "rainfall_may_june_mm",
    "temperature_annual_c",
    "temperature_may_c",
    "temperature_june_c",
    "temperature_may_june_c"
]

categorical_features = ["district"]

required_cols = ["district", "year", target] + numeric_features[1:]

missing = [c for c in required_cols if c not in df.columns]
if missing:
    raise ValueError(f"Missing columns: {missing}")

data = df.copy()
data["district"] = data["district"].astype(str).str.strip().str.upper()
data["year"] = pd.to_numeric(data["year"], errors="raise")
data[target] = pd.to_numeric(data[target], errors="raise")

data = data.dropna(subset=[target, "district", "year"]).copy()
data["year"] = data["year"].astype(int)

X = data[numeric_features + categorical_features]
y = data[target]

# ------------------------------------------------------------
# 2. MODELS AND PREPROCESSING
# ------------------------------------------------------------

numeric_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("num", numeric_transformer, numeric_features),
    ("cat", categorical_transformer, categorical_features)
])

models = {
    "Ridge": Ridge(alpha=1.0),
    "Random Forest": RandomForestRegressor(
        n_estimators=300,
        random_state=42,
        min_samples_leaf=2,
        n_jobs=-1
    ),
    "XGBoost": XGBRegressor(
        n_estimators=300,
        max_depth=3,
        learning_rate=0.05,
        subsample=0.9,
        colsample_bytree=0.9,
        objective="reg:squarederror",
        random_state=42,
        n_jobs=-1
    )
}

prediction_rows = []
metric_rows = []

# ------------------------------------------------------------
# 3. RANDOM 70/30 SPLIT
# ------------------------------------------------------------

train_idx, test_idx = train_test_split(
    data.index, test_size=0.30, random_state=42
)

for model_name, model in models.items():
    pipe = Pipeline([
        ("preprocess", preprocessor),
        ("model", model)
    ])

    pipe.fit(X.loc[train_idx], y.loc[train_idx])
    pred = pipe.predict(X.loc[test_idx])

    for idx, actual, predicted in zip(test_idx, y.loc[test_idx], pred):
        prediction_rows.append({
            "district": data.loc[idx, "district"],
            "year": int(data.loc[idx, "year"]),
            "actual": float(actual),
            "predicted": float(predicted),
            "model": model_name,
            "validation": "random",
            "feature_version": "With year"
        })

    metric_rows.append({
        "model": model_name,
        "validation": "random",
        "feature_version": "With year",
        "R2": r2_score(y.loc[test_idx], pred),
        "MAE": mean_absolute_error(y.loc[test_idx], pred),
        "RMSE": np.sqrt(mean_squared_error(y.loc[test_idx], pred)),
        "n": len(test_idx)
    })

# ------------------------------------------------------------
# 4. EXPANDING-WINDOW WALK-FORWARD VALIDATION
# Train only on years before each test year
# ------------------------------------------------------------

for model_name, model in models.items():
    for test_year in sorted(data["year"].unique()):
        if test_year < 2005:
            continue

        train_mask = data["year"] < test_year
        test_mask = data["year"] == test_year

        if train_mask.sum() == 0 or test_mask.sum() == 0:
            continue

        pipe = Pipeline([
            ("preprocess", preprocessor),
            ("model", model)
        ])

        pipe.fit(X.loc[train_mask], y.loc[train_mask])
        pred = pipe.predict(X.loc[test_mask])

        for idx, actual, predicted in zip(
            data.index[test_mask], y.loc[test_mask], pred
        ):
            prediction_rows.append({
                "district": data.loc[idx, "district"],
                "year": int(data.loc[idx, "year"]),
                "actual": float(actual),
                "predicted": float(predicted),
                "model": model_name,
                "validation": "walk_forward",
                "feature_version": "With year"
            })

        metric_rows.append({
            "model": model_name,
            "validation": "walk_forward",
            "feature_version": "With year",
            "R2": r2_score(y.loc[test_mask], pred),
            "MAE": mean_absolute_error(y.loc[test_mask], pred),
            "RMSE": np.sqrt(mean_squared_error(y.loc[test_mask], pred)),
            "n": int(test_mask.sum())
        })

# ------------------------------------------------------------
# 5. SAVE RESULTS IN MEMORY
# ------------------------------------------------------------

predictions = pd.DataFrame(prediction_rows)
metrics = pd.DataFrame(metric_rows)

print("Predictions recreated:", predictions.shape)
print("\nMetrics:")
display(metrics.round(3))

print("\nPrediction columns:")
print(predictions.columns.tolist())

# Save CSV copies for safety
predictions.to_csv("baseline_predictions_recreated.csv", index=False)
metrics.to_csv("baseline_metrics_recreated.csv", index=False)

print("\nSaved baseline_predictions_recreated.csv")
print("Saved baseline_metrics_recreated.csv")
```

```python

# ============================================================
# IRRIGATION SUBGROUP ANALYSIS
# ============================================================

import pandas as pd
import numpy as np
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

# 1. Keep only genuine observed irrigation records
valid_groups = ["Irrigated", "Mixed", "Rain-fed"]

irrigation = df.copy()

irrigation["year"] = pd.to_numeric(irrigation["year"], errors="coerce")
irrigation["irrigation_source_year"] = pd.to_numeric(
    irrigation["irrigation_source_year"], errors="coerce"
)

irrigation = irrigation[
    irrigation["irrigation_group"].isin(valid_groups)
    & (irrigation["year"] == irrigation["irrigation_source_year"])
].copy()

irrigation["district"] = (
    irrigation["district"].astype(str).str.strip().str.upper()
)

print("Observed irrigation records:", len(irrigation))
print("\nObserved irrigation categories:")
print(irrigation["irrigation_group"].value_counts())

print("\nObserved irrigation years:")
print(sorted(irrigation["year"].dropna().unique()))

# 2. Attach irrigation categories to predictions
pred = predictions.copy()
pred["district"] = pred["district"].astype(str).str.strip().str.upper()
pred["year"] = pd.to_numeric(pred["year"], errors="coerce")

pred = pred.merge(
    irrigation[
        ["district", "year", "irrigation_group", "irrigation_source_year"]
    ].drop_duplicates(["district", "year"]),
    on=["district", "year"],
    how="inner"
)

print("\nPrediction rows with observed irrigation:", len(pred))

# 3. Calculate metrics for each irrigation group
results = []

for keys, group in pred.groupby(
    ["feature_version", "model", "validation", "irrigation_group"]
):
    feature_version, model, validation, irrigation_group = keys

    # R² is not meaningful for groups with fewer than two observations
    r2 = (
        r2_score(group["actual"], group["predicted"])
        if len(group) >= 2 else np.nan
    )

    results.append({
        "feature_version": feature_version,
        "model": model,
        "validation": validation,
        "irrigation_group": irrigation_group,
        "n": len(group),
        "R2": r2,
        "MAE": mean_absolute_error(group["actual"], group["predicted"]),
        "RMSE": np.sqrt(
            mean_squared_error(group["actual"], group["predicted"])
        )
    })

irrigation_metrics = pd.DataFrame(results)

print("\nIRRIGATION SUBGROUP METRICS")
display(irrigation_metrics.round(3))

# 4. Save outputs
irrigation_metrics.to_csv("irrigation_subgroup_metrics.csv", index=False)
pred.to_csv("irrigation_subgroup_predictions.csv", index=False)

print("\nSaved:")
print("- irrigation_subgroup_metrics.csv")
print("- irrigation_subgroup_predictions.csv")
```

```python

import pandas as pd
import numpy as np
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.linear_model import Ridge
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

# Install XGBoost if needed
%pip -q install xgboost
from xgboost import XGBRegressor

# Load dataset
filename = "karnataka_rice_climate_irrigation_merged_2000_2023.csv"
df = pd.read_csv(filename)
df["district"] = df["district"].astype(str).str.strip().str.upper()

target = "Yield (Kg/ha)"
num_cols = [
    "year", "rainfall_annual_mm", "rainfall_may_mm",
    "rainfall_june_mm", "rainfall_may_june_mm",
    "temperature_annual_c", "temperature_may_c",
    "temperature_june_c", "temperature_may_june_c"
]
cat_cols = ["district"]

df["year"] = pd.to_numeric(df["year"], errors="coerce")
df[target] = pd.to_numeric(df[target], errors="coerce")
df = df.dropna(subset=[target, "year", "district"]).copy()
df["year"] = df["year"].astype(int)

preprocess = ColumnTransformer([
    ("num", Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]), num_cols),
    ("cat", Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(handle_unknown="ignore"))
    ]), cat_cols)
])

models = {
    "Ridge": Ridge(alpha=1.0),
    "Random Forest": RandomForestRegressor(
        n_estimators=300, min_samples_leaf=2,
        random_state=42, n_jobs=-1
    ),
    "XGBoost": XGBRegressor(
        n_estimators=300, max_depth=3, learning_rate=0.05,
        subsample=0.9, colsample_bytree=0.9,
        objective="reg:squarederror", random_state=42, n_jobs=-1
    )
}

pred_rows = []
metric_rows = []
X = df[num_cols + cat_cols]
y = df[target]

# Random 70/30 split
train_idx, test_idx = train_test_split(
    df.index, test_size=0.30, random_state=42
)

for name, model in models.items():
    pipe = Pipeline([("prep", preprocess), ("model", model)])
    pipe.fit(X.loc[train_idx], y.loc[train_idx])
    yp = pipe.predict(X.loc[test_idx])

    for idx, actual, predicted in zip(test_idx, y.loc[test_idx], yp):
        pred_rows.append({
            "district": df.loc[idx, "district"],
            "year": int(df.loc[idx, "year"]),
            "actual": float(actual), "predicted": float(predicted),
            "model": name, "validation": "random",
            "feature_version": "With year"
        })

    metric_rows.append({
        "model": name, "validation": "random",
        "R2": r2_score(y.loc[test_idx], yp),
        "MAE": mean_absolute_error(y.loc[test_idx], yp),
        "RMSE": np.sqrt(mean_squared_error(y.loc[test_idx], yp))
    })

# Expanding-window walk-forward predictions, starting in 2005
for name, model in models.items():
    for yr in sorted(df["year"].unique()):
        if yr < 2005:
            continue

        train_mask = df["year"] < yr
        test_mask = df["year"] == yr
        if not train_mask.any() or not test_mask.any():
            continue

        pipe = Pipeline([("prep", preprocess), ("model", model)])
        pipe.fit(X.loc[train_mask], y.loc[train_mask])
        yp = pipe.predict(X.loc[test_mask])
        ids = df.index[test_mask]

        for idx, actual, predicted in zip(ids, y.loc[ids], yp):
            pred_rows.append({
                "district": df.loc[idx, "district"],
                "year": int(df.loc[idx, "year"]),
                "actual": float(actual), "predicted": float(predicted),
                "model": name, "validation": "walk_forward",
                "feature_version": "With year"
            })

        metric_rows.append({
            "model": name, "validation": f"walk_forward_{yr}",
            "R2": r2_score(y.loc[test_mask], yp) if len(ids) >= 2 else np.nan,
            "MAE": mean_absolute_error(y.loc[test_mask], yp),
            "RMSE": np.sqrt(mean_squared_error(y.loc[test_mask], yp))
        })

predictions = pd.DataFrame(pred_rows)
metrics = pd.DataFrame(metric_rows)

print("Predictions recreated:", predictions.shape)
print(predictions.groupby(["model", "validation"]).size())
```

```python
# Install XGBoost if needed
!pip -q install xgboost

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.linear_model import Ridge
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
from xgboost import XGBRegressor

# Target
target = "Yield (Kg/ha)"

# Predictors: district, year, and climate only
numeric_features = [
    "year",
    "rainfall_annual_mm",
    "rainfall_may_mm",
    "rainfall_june_mm",
    "rainfall_may_june_mm",
    "temperature_annual_c",
    "temperature_may_c",
    "temperature_june_c",
    "temperature_may_june_c"
]

categorical_features = ["district"]
features = numeric_features + categorical_features

X = df[features].copy()
y = df[target].copy()

# Models
models = {
    "Ridge": Ridge(alpha=1.0),
    "Random Forest": RandomForestRegressor(
        n_estimators=300,
        random_state=42,
        min_samples_leaf=2
    ),
    "XGBoost": XGBRegressor(
        n_estimators=300,
        max_depth=3,
        learning_rate=0.03,
        subsample=0.9,
        colsample_bytree=0.9,
        objective="reg:squarederror",
        random_state=42
    )
}

# Preprocessing is fitted only on training data
def make_pipeline(model):
    numeric_pipe = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ])

    categorical_pipe = Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(handle_unknown="ignore"))
    ])

    preprocessor = ColumnTransformer([
        ("numeric", numeric_pipe, numeric_features),
        ("categorical", categorical_pipe, categorical_features)
    ])

    return Pipeline([
        ("preprocessor", preprocessor),
        ("model", model)
    ])

# Random 70/30 split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=42
)

results = []
random_predictions = {}

for name, model in models.items():
    pipeline = make_pipeline(model)
    pipeline.fit(X_train, y_train)
    predictions = pipeline.predict(X_test)

    random_predictions[name] = predictions

    results.append({
        "Validation": "Random split",
        "Model": name,
        "R2": r2_score(y_test, predictions),
        "MAE": mean_absolute_error(y_test, predictions),
        "RMSE": np.sqrt(mean_squared_error(y_test, predictions))
    })

random_results = pd.DataFrame(results)
display(random_results.round(3))
```

```python
# Chronological expanding-window walk-forward validation

walk_predictions = []
observed_years = sorted(df["year"].unique())

for test_year in observed_years[1:]:
    train_mask = df["year"] < test_year
    test_mask = df["year"] == test_year

    X_train_fold = X.loc[train_mask]
    y_train_fold = y.loc[train_mask]
    X_test_fold = X.loc[test_mask]
    y_test_fold = y.loc[test_mask]

    print(
        f"Training through {X_train_fold['year'].max()} "
        f"-> testing {test_year} "
        f"({len(X_train_fold)} training rows, "
        f"{len(X_test_fold)} test rows)"
    )

    for name, model in models.items():
        pipeline = make_pipeline(model)
        pipeline.fit(X_train_fold, y_train_fold)
        predictions = pipeline.predict(X_test_fold)

        for actual, predicted in zip(y_test_fold, predictions):
            walk_predictions.append({
                "year": test_year,
                "Model": name,
                "Actual_Yield": actual,
                "Predicted_Yield": predicted
            })

walk_df = pd.DataFrame(walk_predictions)

# Calculate overall metrics across all future-year predictions
walk_results = []

for name, group in walk_df.groupby("Model"):
    actual = group["Actual_Yield"]
    predicted = group["Predicted_Yield"]

    walk_results.append({
        "Validation": "Walk-forward",
        "Model": name,
        "R2": r2_score(actual, predicted),
        "MAE": mean_absolute_error(actual, predicted),
        "RMSE": np.sqrt(mean_squared_error(actual, predicted))
    })

walk_results = pd.DataFrame(walk_results)
display(walk_results.round(3))
```

```python
# Examine walk-forward performance year by year

yearly_results = []

for (year, model_name), group in walk_df.groupby(["year", "Model"]):
    actual = group["Actual_Yield"]
    predicted = group["Predicted_Yield"]

    yearly_results.append({
        "Year": year,
        "Model": model_name,
        "Test_rows": len(group),
        "R2": r2_score(actual, predicted) if len(group) >= 2 else np.nan,
        "MAE": mean_absolute_error(actual, predicted),
        "RMSE": np.sqrt(mean_squared_error(actual, predicted)),
        "Actual_mean": actual.mean(),
        "Predicted_mean": predicted.mean(),
        "Min_prediction": predicted.min(),
        "Max_prediction": predicted.max()
    })

yearly_results = pd.DataFrame(yearly_results)

display(
    yearly_results
    .sort_values(["Year", "Model"])
    .round(2)
    .reset_index(drop=True)
)
```

```python

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.linear_model import Ridge
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor

# Preprocessing for numeric predictors
numeric_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Preprocessing for district names
categorical_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("numeric", numeric_transformer, numeric_features),
    ("district", categorical_transformer, categorical_features)
])

# Define the models
models = {
    "Ridge": Ridge(alpha=1.0),

    "Random Forest": RandomForestRegressor(
        n_estimators=300,
        random_state=42,
        min_samples_leaf=2
    ),

    "XGBoost": XGBRegressor(
        n_estimators=300,
        max_depth=3,
        learning_rate=0.03,
        subsample=0.9,
        colsample_bytree=0.9,
        objective="reg:squarederror",
        random_state=42
    )
}

# Build a separate pipeline for each model
pipelines = {
    name: Pipeline([
        ("preprocessor", preprocessor),
        ("model", model)
    ])
    for name, model in models.items()
}

print("Models prepared:")
for name in pipelines:
    print("-", name)
```

```python
# ============================================================
# CORRECTED FIVE-OBSERVED-YEAR IRRIGATION SUBGROUP EXPERIMENT
# Run in Google Colab. This retrains models from the audited
# merged dataset; it does NOT reuse the old prediction CSV.
# ============================================================

from google.colab import files
import pandas as pd
import numpy as np

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
from sklearn.linear_model import Ridge
from sklearn.ensemble import RandomForestRegressor

# ----------------------------
# 1. Upload audited merged data
# ----------------------------
uploaded = files.upload()
csv_names = [n for n in uploaded if n.lower().endswith(".csv")]
if not csv_names:
    raise ValueError("Upload the audited merged irrigation CSV.")
# If multiple CSVs were uploaded, choose the full merged panel:
candidates = [n for n in csv_names if "merged" in n.lower() and "irrigation" in n.lower()]
filename = candidates[0] if candidates else csv_names[0]
data = pd.read_csv(filename)
print("Loaded:", filename, "| shape:", data.shape)

# ----------------------------
# 2. Check column names
# ----------------------------
required = [
    "district", "year", "Yield (Kg/ha)", "irrigated_share_pct",
    "irrigation_group", "irrigation_match_status"
]
missing = [c for c in required if c not in data.columns]
if missing:
    raise ValueError(
        f"Missing columns: {missing}\n"
        "Please upload karnataka_rice_climate_irrigation_merged_2000_2023_5YEAR_AUDITED.csv"
    )

# Climate predictors available in the merged panel
climate_candidates = [
    "rainfall_annual_mm", "rainfall_may_mm", "rainfall_june_mm",
    "rainfall_may_june_mm", "temperature_annual_c", "temperature_may_c",
    "temperature_june_c", "temperature_may_june_c"
]
climate_cols = [c for c in climate_candidates if c in data.columns]
if not climate_cols:
    raise ValueError("No expected climate columns were found.")

# Use only directly observed irrigation records, not legacy fallback labels.
observed = data[
    data["irrigated_share_pct"].notna()
    & data["irrigation_group"].isin(["Irrigated", "Mixed", "Rain-fed"])
].copy()

observed["year"] = pd.to_numeric(observed["year"], errors="coerce")
observed["Yield (Kg/ha)"] = pd.to_numeric(observed["Yield (Kg/ha)"], errors="coerce")
observed = observed.dropna(subset=["district", "year", "Yield (Kg/ha)"])
observed = observed[observed["year"].isin([2001, 2002, 2009, 2011, 2017])].copy()

# Never use area or production as predictors: yield is derived from these.
# Do not use irrigated_share_pct as a predictor in this experiment:
# it defines the subgroup labels and would make the subgroup feature circular.
base_features = ["district", "irrigation_group"] + climate_cols
categorical_cols = ["district", "irrigation_group"]
numeric_cols = climate_cols

print("\nObserved irrigation rows:", len(observed))
print("Years:", sorted(observed["year"].astype(int).unique()))
print("Group counts:")
print(observed["irrigation_group"].value_counts())
print("\nClimate predictors:", climate_cols)
print("Excluded predictors: Area (Ha), Production (Tonnes), Yield (Kg/ha), irrigated_share_pct")

# ----------------------------
# 3. Models and leakage-safe preprocessing
# ----------------------------
try:
    from xgboost import XGBRegressor
    xgb_available = True
except ImportError:
    xgb_available = False
    print("\nXGBoost is not installed. Ridge and Random Forest will run; "
          "you can install xgboost separately if needed.")

def make_model(model_name, include_year):
    features = base_features + (["year"] if include_year else [])
    cats = categorical_cols
    nums = numeric_cols + (["year"] if include_year else [])

    preprocessor = ColumnTransformer(
        transformers=[
            ("num", Pipeline([
                ("imputer", SimpleImputer(strategy="median")),
                ("scaler", StandardScaler())
            ]), nums),
            ("cat", Pipeline([
                ("imputer", SimpleImputer(strategy="most_frequent")),
                ("onehot", OneHotEncoder(handle_unknown="ignore"))
            ]), cats)
        ],
        remainder="drop"
    )

    if model_name == "Ridge":
        estimator = Ridge(alpha=1.0)
    elif model_name == "Random Forest":
        estimator = RandomForestRegressor(
            n_estimators=300, random_state=42, min_samples_leaf=2,
            n_jobs=-1
        )
    elif model_name == "XGBoost":
        if not xgb_available:
            return None
        estimator = XGBRegressor(
            n_estimators=300, max_depth=3, learning_rate=0.03,
            subsample=0.9, colsample_bytree=0.9,
            objective="reg:squarederror", random_state=42,
            n_jobs=-1
        )
    else:
        raise ValueError(model_name)

    return Pipeline([("preprocess", preprocessor), ("model", estimator)])

model_names = ["Ridge", "Random Forest", "XGBoost"]
if not xgb_available:
    model_names = ["Ridge", "Random Forest"]

def score_predictions(y_true, y_pred):
    return {
        "n": len(y_true),
        "R2": r2_score(y_true, y_pred) if len(y_true) >= 2 else np.nan,
        "MAE": mean_absolute_error(y_true, y_pred),
        "RMSE": np.sqrt(mean_squared_error(y_true, y_pred))
    }

all_predictions = []
all_metrics = []

# ----------------------------
# 4. Random 70/30 split
# This is a comparison only; it can mix years across train/test.
# ----------------------------
for include_year in [False, True]:
    feature_set = base_features + (["year"] if include_year else [])
    train_idx, test_idx = train_test_split(
        observed.index, test_size=0.30, random_state=42
    )
    train = observed.loc[train_idx]
    test = observed.loc[test_idx]

    for model_name in model_names:
        model = make_model(model_name, include_year)
        model.fit(train[feature_set], train["Yield (Kg/ha)"])
        pred = model.predict(test[feature_set])

        for group in ["Irrigated", "Mixed", "Rain-fed"]:
            mask = test["irrigation_group"].to_numpy() == group
            if mask.sum() == 0:
                continue
            met = score_predictions(test["Yield (Kg/ha)"].to_numpy()[mask], pred[mask])
            all_metrics.append({
                "validation": "Random 70/30",
                "year_feature": include_year,
                "model": model_name,
                "irrigation_group": group,
                **met
            })

            for pos in np.where(mask)[0]:
                row = test.iloc[pos]
                all_predictions.append({
                    "validation": "Random 70/30",
                    "year_feature": include_year,
                    "model": model_name,
                    "district": row["district"],
                    "year": int(row["year"]),
                    "irrigation_group": row["irrigation_group"],
                    "actual_yield": row["Yield (Kg/ha)"],
                    "predicted_yield": pred[pos]
                })

# ----------------------------
# 5. Expanding walk-forward
# Only train on years strictly earlier than the test year.
# With these sparse years, the first fold is very small; interpret
# results cautiously. Fold metrics are also saved separately.
# ----------------------------
for include_year in [False, True]:
    feature_set = base_features + (["year"] if include_year else [])

    for test_year in [2002, 2009, 2011, 2017]:
        train = observed[observed["year"] < test_year].copy()
        test = observed[observed["year"] == test_year].copy()

        if train.empty or test.empty:
            continue

        print(f"Walk-forward fold {test_year}: train n={len(train)}, test n={len(test)}, "
              f"training years={sorted(train['year'].astype(int).unique())}")

        for model_name in model_names:
            model = make_model(model_name, include_year)
            model.fit(train[feature_set], train["Yield (Kg/ha)"])
            pred = model.predict(test[feature_set])

            for group in ["Irrigated", "Mixed", "Rain-fed"]:
                mask = test["irrigation_group"].to_numpy() == group
                if mask.sum() == 0:
                    continue
                met = score_predictions(test["Yield (Kg/ha)"].to_numpy()[mask], pred[mask])
                all_metrics.append({
                    "validation": f"Walk-forward test year {test_year}",
                    "year_feature": include_year,
                    "model": model_name,
                    "irrigation_group": group,
                    **met
                })

                for pos in np.where(mask)[0]:
                    row = test.iloc[pos]
                    all_predictions.append({
                        "validation": f"Walk-forward test year {test_year}",
                        "year_feature": include_year,
                        "model": model_name,
                        "district": row["district"],
                        "year": int(row["year"]),
                        "irrigation_group": row["irrigation_group"],
                        "actual_yield": row["Yield (Kg/ha)"],
                        "predicted_yield": pred[pos]
                    })

# ----------------------------
# 6. Save outputs and download
# ----------------------------
metrics_df = pd.DataFrame(all_metrics)
predictions_df = pd.DataFrame(all_predictions)

metrics_df.to_csv("corrected_irrigation_five_year_metrics.csv", index=False)
predictions_df.to_csv("corrected_irrigation_five_year_predictions.csv", index=False)

print("\nMETRICS")
display(metrics_df.round(3))
print("\nPrediction rows:", len(predictions_df))
print("Files saved: corrected_irrigation_five_year_metrics.csv")
print("             corrected_irrigation_five_year_predictions.csv")

files.download("corrected_irrigation_five_year_metrics.csv")
files.download("corrected_irrigation_five_year_predictions.csv")
```

```python
# ============================================================
# FIVE OBSERVED-YEAR IRRIGATION SUBGROUP EXPERIMENT — V2
# Run in Google Colab. Upload ONLY the audited merged panel CSV.
# This script fits models in your Colab session; ChatGPT is not
# running the models. It adds strict input checks to prevent the
# wrong CSV being silently selected when multiple CSVs are uploaded.
# ============================================================

from google.colab import files
import pandas as pd
import numpy as np
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
from sklearn.linear_model import Ridge
from sklearn.ensemble import RandomForestRegressor

EXPECTED_YEARS = [2001, 2002, 2009, 2011, 2017]
EXPECTED_FILENAME_HINT = "karnataka_rice_climate_irrigation_merged_2000_2023_5YEAR_AUDITED"

print("Upload ONLY this file: karnataka_rice_climate_irrigation_merged_2000_2023_5YEAR_AUDITED (4).csv")
uploaded = files.upload()
csv_names = [n for n in uploaded if n.lower().endswith('.csv')]
if len(csv_names) != 1:
    raise ValueError(f"Please rerun and upload exactly ONE CSV (the audited merged panel). Uploaded CSVs: {csv_names}")
filename = csv_names[0]
if 'merged' not in filename.lower() or 'irrigation' not in filename.lower():
    raise ValueError(f"Wrong file selected: {filename}. Upload the audited merged irrigation panel, not a metrics/predictions CSV.")

data = pd.read_csv(filename)
required = ['district','year','Yield (Kg/ha)','irrigated_share_pct','irrigation_group','irrigation_match_status']
missing = [c for c in required if c not in data.columns]
if missing:
    raise ValueError(f"Missing required columns: {missing}. This is probably not the audited merged panel.")
print(f"\nLoaded {filename}; shape={data.shape}")

# Audit source panel before modeling.
data['year'] = pd.to_numeric(data['year'], errors='coerce')
data['irrigated_share_pct'] = pd.to_numeric(data['irrigated_share_pct'], errors='coerce')
obs = data[data['irrigated_share_pct'].notna() & data['irrigation_group'].isin(['Irrigated','Mixed','Rain-fed'])].copy()
print("\nAUDIT: directly observed irrigation records by year (before yield filtering)")
print(obs.groupby('year').agg(rows=('district','size'), districts=('district','nunique')).to_string())
observed_years = sorted(obs['year'].dropna().astype(int).unique().tolist())
print('Observed years found:', observed_years)
if observed_years != EXPECTED_YEARS:
    raise ValueError(
        f"Expected observed irrigation years {EXPECTED_YEARS}, but found {observed_years}. "
        "Stop here: check the uploaded source file rather than interpreting model metrics."
    )

obs['Yield (Kg/ha)'] = pd.to_numeric(obs['Yield (Kg/ha)'], errors='coerce')
obs = obs.dropna(subset=['district','year','Yield (Kg/ha)']).copy()
obs['year'] = obs['year'].astype(int)
print('\nRows available for modeling by year (after requiring yield):')
print(obs.groupby('year').agg(rows=('district','size'), districts=('district','nunique')).to_string())

climate_candidates = [
    'rainfall_annual_mm','rainfall_may_mm','rainfall_june_mm','rainfall_may_june_mm',
    'temperature_annual_c','temperature_may_c','temperature_june_c','temperature_may_june_c'
]
climate_cols = [c for c in climate_candidates if c in obs.columns]
if not climate_cols:
    raise ValueError('No expected climate predictor columns found.')
base_features = ['district','irrigation_group'] + climate_cols
categorical_cols = ['district','irrigation_group']
numeric_cols = climate_cols
print('\nClimate predictors:', climate_cols)
print('Excluded from predictors: Area (Ha), Production (Tonnes), Yield (Kg/ha), irrigated_share_pct')
print('Irrigation group is used for subgroup labels and as a categorical predictor; irrigated_share_pct itself is excluded.')

try:
    from xgboost import XGBRegressor
    xgb_available = True
except ImportError:
    xgb_available = False
    print('XGBoost unavailable; running Ridge and Random Forest only.')

model_names = ['Ridge','Random Forest'] + (['XGBoost'] if xgb_available else [])

def make_model(model_name, include_year):
    nums = numeric_cols + (['year'] if include_year else [])
    pre = ColumnTransformer([
        ('num', Pipeline([('imputer', SimpleImputer(strategy='median')), ('scaler', StandardScaler())]), nums),
        ('cat', Pipeline([('imputer', SimpleImputer(strategy='most_frequent')), ('onehot', OneHotEncoder(handle_unknown='ignore'))]), categorical_cols)
    ], remainder='drop')
    if model_name == 'Ridge':
        estimator = Ridge(alpha=1.0)
    elif model_name == 'Random Forest':
        estimator = RandomForestRegressor(n_estimators=300, random_state=42, min_samples_leaf=2, n_jobs=-1)
    else:
        estimator = XGBRegressor(n_estimators=300, max_depth=3, learning_rate=0.03, subsample=0.9,
                                 colsample_bytree=0.9, objective='reg:squarederror', random_state=42, n_jobs=-1)
    return Pipeline([('preprocess', pre), ('model', estimator)])

def score(y, p):
    return {'n': len(y), 'R2': r2_score(y,p) if len(y)>=2 else np.nan,
            'MAE': mean_absolute_error(y,p), 'RMSE': np.sqrt(mean_squared_error(y,p))}

metrics, predictions = [], []
# Random 70/30 split: comparison only; training and test may contain different years.
train_idx, test_idx = train_test_split(obs.index, test_size=0.30, random_state=42)
train, test = obs.loc[train_idx].copy(), obs.loc[test_idx].copy()
print('\nRandom split years in test set:', sorted(test['year'].unique().tolist()))
for include_year in [False, True]:
    feats = base_features + (['year'] if include_year else [])
    for model_name in model_names:
        model = make_model(model_name, include_year)
        model.fit(train[feats], train['Yield (Kg/ha)'])
        pred = model.predict(test[feats])
        for group in ['Irrigated','Mixed','Rain-fed']:
            mask = test['irrigation_group'].to_numpy() == group
            if mask.sum() == 0: continue
            metrics.append({'validation':'Random 70/30','year_feature':include_year,'model':model_name,'irrigation_group':group,
                            **score(test['Yield (Kg/ha)'].to_numpy()[mask], pred[mask])})
        for pos, (_, row) in enumerate(test.iterrows()):
            predictions.append({'validation':'Random 70/30','year_feature':include_year,'model':model_name,
                                'district':row['district'],'year':int(row['year']),'irrigation_group':row['irrigation_group'],
                                'actual_yield':row['Yield (Kg/ha)'],'predicted_yield':pred[pos]})

# Expanding walk-forward. 2001 is the first observed year, so it is not a test fold;
# 2002 is the first possible fold, trained only on 2001 records.
for include_year in [False, True]:
    feats = base_features + (['year'] if include_year else [])
    for test_year in [2002,2009,2011,2017]:
        train = obs[obs['year'] < test_year].copy()
        test = obs[obs['year'] == test_year].copy()
        if train.empty or test.empty:
            print(f'SKIPPED fold {test_year}: train n={len(train)}, test n={len(test)}')
            continue
        print(f'Walk-forward fold {test_year}: train n={len(train)} ({sorted(train.year.unique())}), test n={len(test)}')
        for model_name in model_names:
            model = make_model(model_name, include_year)
            model.fit(train[feats], train['Yield (Kg/ha)'])
            pred = model.predict(test[feats])
            for group in ['Irrigated','Mixed','Rain-fed']:
                mask = test['irrigation_group'].to_numpy() == group
                if mask.sum() == 0: continue
                metrics.append({'validation':f'Walk-forward test year {test_year}','year_feature':include_year,
                                'model':model_name,'irrigation_group':group,
                                **score(test['Yield (Kg/ha)'].to_numpy()[mask],pred[mask])})
            for pos, (_, row) in enumerate(test.iterrows()):
                predictions.append({'validation':f'Walk-forward test year {test_year}','year_feature':include_year,
                                    'model':model_name,'district':row['district'],'year':int(row['year']),
                                    'irrigation_group':row['irrigation_group'],'actual_yield':row['Yield (Kg/ha)'],
                                    'predicted_yield':pred[pos]})

metrics_df = pd.DataFrame(metrics)
pred_df = pd.DataFrame(predictions)
metrics_file = 'corrected_irrigation_five_year_metrics_v2.csv'
pred_file = 'corrected_irrigation_five_year_predictions_v2.csv'
metrics_df.to_csv(metrics_file,index=False)
pred_df.to_csv(pred_file,index=False)
print('\nMETRICS')
display(metrics_df.round(3))
print('\nPrediction counts by validation:')
print(pred_df.groupby(['validation','year_feature','model']).size().to_string())
print('\nPrediction years:', sorted(pred_df['year'].unique().tolist()))
print('Saved:', metrics_file, 'and', pred_file)
files.download(metrics_file)
files.download(pred_file)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.linear_model import Ridge
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor

def make_pipeline(model, feature_columns):
    categorical = [
        col for col in feature_columns
        if df_exp[col].dtype == "object"
    ]
    numerical = [
        col for col in feature_columns
        if col not in categorical
    ]

    numeric_pipeline = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ])

    categorical_pipeline = Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(handle_unknown="ignore"))
    ])

    preprocessing = ColumnTransformer([
        ("numeric", numeric_pipeline, numerical),
        ("categorical", categorical_pipeline, categorical)
    ])

    return Pipeline([
        ("preprocessing", preprocessing),
        ("model", model)
    ])

models = {
    "Ridge": Ridge(alpha=1.0),
    "Random Forest": RandomForestRegressor(
        n_estimators=300,
        random_state=42,
        n_jobs=-1
    ),
    "XGBoost": XGBRegressor(
        n_estimators=300,
        max_depth=3,
        learning_rate=0.05,
        objective="reg:squarederror",
        random_state=42,
        n_jobs=-1
    )
}

print("Models prepared:", list(models.keys()))
print("One-hot encoding: configured")
print("Preprocessing and models are ready.")
```

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
import pandas as pd

results = []

for model_name, model_template in models.items():
    for condition, feature_columns in [
        ("Without PMFBY flag", features_without_pmfby),
        ("With PMFBY flag", features_with_pmfby)
    ]:
        # Use the same train/test rows for every comparison
        X_train = df_exp.loc[train_idx, feature_columns]
        X_test = df_exp.loc[test_idx, feature_columns]
        y_train = df_exp.loc[train_idx, TARGET]
        y_test = df_exp.loc[test_idx, TARGET]

        # Create a fresh pipeline for each model and condition
        pipeline = make_pipeline(
            model_template.__class__(**model_template.get_params()),
            feature_columns
        )

        pipeline.fit(X_train, y_train)
        predictions = pipeline.predict(X_test)

        results.append({
            "Model": model_name,
            "Condition": condition,
            "R2": r2_score(y_test, predictions),
            "MAE": mean_absolute_error(y_test, predictions),
            "RMSE": mean_squared_error(y_test, predictions) ** 0.5
        })

pmfby_results = pd.DataFrame(results)

print("PMFBY FLAG COMPARISON RESULTS")
display(
    pmfby_results.round(3).sort_values(
        ["Model", "Condition"]
    ).reset_index(drop=True)
)
```

```python
import numpy as np
import pandas as pd
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

baseline_predictions = []

for test_year in range(2005, 2024):
    train_data = df_exp[df_exp["year"] < test_year].copy()
    test_data = df_exp[df_exp["year"] == test_year].copy()

    # Historical average yield per district using past years only
    district_means = train_data.groupby("district")["Yield (Kg/ha)"].mean()
    overall_mean = train_data["Yield (Kg/ha)"].mean()

    # If a district has no earlier records, use the training-period mean
    predictions = test_data["district"].map(district_means).fillna(overall_mean)

    for district, actual, predicted in zip(
        test_data["district"],
        test_data["Yield (Kg/ha)"],
        predictions
    ):
        baseline_predictions.append({
            "Test_Year": test_year,
            "District": district,
            "Actual_Yield": actual,
            "Predicted_Yield": predicted
        })

baseline_df = pd.DataFrame(baseline_predictions)

baseline_results = pd.DataFrame([{
    "Model": "District historical-mean baseline",
    "Test_Observations": len(baseline_df),
    "R2": r2_score(
        baseline_df["Actual_Yield"],
        baseline_df["Predicted_Yield"]
    ),
    "MAE": mean_absolute_error(
        baseline_df["Actual_Yield"],
        baseline_df["Predicted_Yield"]
    ),
    "RMSE": np.sqrt(mean_squared_error(
        baseline_df["Actual_Yield"],
        baseline_df["Predicted_Yield"]
    ))
}])

display(baseline_results.round(3))
```

```python
# Combine the baseline and walk-forward model results

baseline_row = baseline_results.copy()

walk_rows = overall_walk_df[
    ["Model", "Condition", "Test_Observations", "R2", "MAE", "RMSE"]
].copy()

baseline_row["Condition"] = "Baseline"
walk_rows["Condition"] = walk_rows["Condition"].astype(str)

baseline_comparison = pd.concat(
    [baseline_row, walk_rows],
    ignore_index=True
)

baseline_comparison.to_csv(
    "/content/walk_forward_vs_baseline.csv",
    index=False
)

display(baseline_comparison.round(3))

print("Saved: walk_forward_vs_baseline.csv")
```
